```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('/Dataset .csv')
```

The Distribution of Price Ranges Among The **Restaurants**

```python
import pandas as pd

df.columns = df.columns.str.strip()

price_ranges_per_counts = (
    df['Price range']
    .value_counts()
    .sort_index()
)

price_ranges_per_counts
```

```python
import matplotlib.pyplot as plt


df.columns = df.columns.str.strip()


price_ranges_per_counts = df['Price range'].value_counts().sort_index()


colors = ['red', 'green', 'yellow', 'blue'][:len(price_ranges_per_counts)]

plt.figure(figsize=(8, 5))
plt.bar(
    price_ranges_per_counts.index.astype(str),
    price_ranges_per_counts.values,
    color=colors
)

plt.title("Distribution of Price Ranges Among Restaurants")
plt.xlabel("Price Ranges")
plt.ylabel("Count")
plt.tight_layout()
plt.show()



```

Calculate The Percentage of Restaurants in Each Price Range **Category**

```python
total_restaurants = len(df)
print("Total number of restaurants:", total_restaurants)
```

```python

df.columns = df.columns.str.strip()


total_restaurants = len(df)


price_ranges_per_counts = df['Price range'].value_counts().sort_index()


percentage_per_price_range = (price_ranges_per_counts / total_restaurants) * 100

print("Percentage of Restaurants in Each Price Range Category:\n")
print(percentage_per_price_range.round(2))
```
