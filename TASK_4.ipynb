```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('/Dataset .csv')
```

Determine The Percentage of Restaurants That Offers Online **Delivery**

```python
total_restaurants = len(df)
print("Total number of restaurants:", total_restaurants)

```

```python
df.columns = df.columns.str.strip()

df_has_online_delivery = df[df['Has Online delivery'] == 'Yes']

total_restaurants = len(df_has_online_delivery)
print("Total restaurants with online delivery:", total_restaurants)
```

```python
total_restaurants = len(df)
online_delivery_count = len(df[df['Has Online delivery'] == 'Yes'])

percentage_online_delivery = (online_delivery_count / total_restaurants) * 100

print(f"Percentage of restaurants offering online delivery: {percentage_online_delivery:.2f}%")
```

The Average Ratings of Restaurants With and Without Online Delivery
**bold text**

```python
import pandas as pd


df.columns = df.columns.str.strip()


df['Aggregate rating'] = pd.to_numeric(df['Aggregate rating'], errors='coerce')


average_rating_with_online_delivery = (
    df[df['Has Online delivery'] == 'Yes']['Aggregate rating'].mean()
)

average_rating_without_online_delivery = (
    df[df['Has Online delivery'] == 'No']['Aggregate rating'].mean()
)

print(f"Average rating of restaurants WITH online delivery: {average_rating_with_online_delivery:.2f}")
print(f"Average rating of restaurants WITHOUT online delivery: {average_rating_without_online_delivery:.2f}")
```

```python
import matplotlib.pyplot as plt

labels = ['With Online Delivery', 'Without Online Delivery']
average_ratings = [
    average_rating_with_online_delivery,
    average_rating_without_online_delivery
]

colors = ['green', 'red']

plt.figure(figsize=(7, 5))
plt.bar(labels, average_ratings, color=colors)

plt.title("Comparison of Average Ratings of Restaurants\nWith and Without Online Delivery")
plt.xlabel("Online Delivery")
plt.ylabel("Average Rating")
plt.ylim(0, 5)  # Rating scale
plt.tight_layout()
plt.show()

```
