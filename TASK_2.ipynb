```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('/Dataset .csv')
```

Identify The City With The Highest Number of **Restaurants**

```python
import pandas as pd



city_counts = df['City'].value_counts()
city_with_the_most_restaurants = city_counts.idxmax()

print("The city with the highest number of restaurants is:", city_with_the_most_restaurants)
```

Calculate The Average Rating For Restaurants in Each **City**

```python
import pandas as pd


df.columns = df.columns.str.strip()

average_rating_of_restaurants_in_each_city = (
    df.groupby('City')['Aggregate rating'].mean()
)

print("Average Rating for Restaurants in Each City:\n")
print(average_rating_of_restaurants_in_each_city)
```

City With The Highest Average **Rating**

```python


import pandas as pd

df.columns = df.columns.str.strip()

average_ratings = df.groupby('City')['Aggregate rating'].mean()

city_with_highest_average_rating = average_ratings.idxmax()

print(f"The city with the highest average rating is {city_with_highest_average_rating}")
```
